```python
import pandas as pd
pd.set_option('compute.use_numexpr', False)
import shap
import joblib
import pandas as pd
import matplotlib.pyplot as plt
```

```python
model = joblib.load(
    "../models/random_forest.joblib"
)

preprocessor = model.named_steps["preprocessor"]

classifier = model.named_steps["classifier"]
```

```python
X_valid = pd.read_csv(
    "../data/processed/X_valid.csv"
)

X_train = pd.read_csv(
    "../data/processed/X_train.csv"
)
```

```python
X_valid_transformed = preprocessor.transform(X_valid)

X_train_transformed = preprocessor.transform(X_train)
```

```python
feature_names = preprocessor.get_feature_names_out()

X_valid_transformed = pd.DataFrame(
    X_valid_transformed.toarray()
    if hasattr(X_valid_transformed, "toarray")
    else X_valid_transformed,
    columns=feature_names
)
```

```python
explainer = shap.TreeExplainer(classifier)

shap_values = explainer.shap_values(
    X_valid_transformed.iloc[:300]
)
```

```python
if isinstance(shap_values, list):
    positive_shap_values = shap_values[1]

elif shap_values.ndim == 3:
    positive_shap_values = shap_values[:, :, 1]

else:
    positive_shap_values = shap_values
```

```python
shap.summary_plot(
    positive_shap_values,
    X_valid_transformed.iloc[:300]
)
```

```python
shap.summary_plot(
    positive_shap_values,
    X_valid_transformed.iloc[:300],
    plot_type="bar"
)
```
